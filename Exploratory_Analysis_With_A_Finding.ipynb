```python
import pandas as pd

file_path = r"C:\Users\MOUNAVI GOLLAPALLI\Downloads\archive (2)\SampleSuperstore.csv"

df = pd.read_csv(file_path, encoding="latin1")

print("Dataset loaded successfully!")
print("Dataset Shape:", df.shape)
display(df.head())

```

```python

print("Dataset Shape:", df.shape)

print("\nColumn Names:")
print(df.columns.tolist())

print("\nDataset Information:")
df.info()

print("\nMissing Values:")
display(df.isnull().sum().to_frame("Missing Values"))

print("\nDuplicate Rows:", df.duplicated().sum())
```

```python

display(df.describe(include="all").T)
```

```python

required_columns = ["Sales", "Profit", "Discount"]

missing_columns = [
    column for column in required_columns
    if column not in df.columns
]

if missing_columns:
    print("Required columns not found:", missing_columns)
else:
    print("Total Sales:", round(df["Sales"].sum(), 2))
    print("Total Profit:", round(df["Profit"].sum(), 2))
    print("Average Discount:", round(df["Discount"].mean(), 3))

    discount_analysis = (
        df.groupby("Discount")
        .agg(
            Orders=("Profit", "count"),
            Total_Sales=("Sales", "sum"),
            Total_Profit=("Profit", "sum"),
            Average_Profit=("Profit", "mean")
        )
        .reset_index()
    )

    display(discount_analysis)
```

```python

plt.figure(figsize=(9, 5))

sns.scatterplot(
    data=df,
    x="Discount",
    y="Profit",
    alpha=0.6
)

plt.axhline(0, linestyle="--")
plt.title("Impact of Discount on Order Profit")
plt.xlabel("Discount")
plt.ylabel("Profit")
plt.tight_layout()
plt.show()
```

```python

category_analysis = (
    df.groupby("Category")
    .agg(
        Total_Sales=("Sales", "sum"),
        Total_Profit=("Profit", "sum"),
        Average_Profit=("Profit", "mean"),
        Total_Orders=("Sales", "count")
    )
    .sort_values("Total_Profit", ascending=False)
    .reset_index()
)

display(category_analysis)
```

```python

fig, axes = plt.subplots(1, 2, figsize=(13, 5))

sns.barplot(
    data=category_analysis,
    x="Category",
    y="Total_Sales",
    ax=axes[0]
)
axes[0].set_title("Total Sales by Category")
axes[0].tick_params(axis="x", rotation=20)

sns.barplot(
    data=category_analysis,
    x="Category",
    y="Total_Profit",
    ax=axes[1]
)
axes[1].set_title("Total Profit by Category")
axes[1].tick_params(axis="x", rotation=20)

plt.tight_layout()
plt.show()
```

```python

region_analysis = (
    df.groupby("Region")
    .agg(
        Total_Sales=("Sales", "sum"),
        Total_Profit=("Profit", "sum"),
        Average_Discount=("Discount", "mean"),
        Order_Count=("Sales", "count")
    )
    .sort_values("Total_Profit", ascending=False)
    .reset_index()
)

display(region_analysis)
```

```python

discount_analysis = (
    df.assign(
        Discount_Group=pd.cut(
            df["Discount"],
            bins=[-0.001, 0, 0.1, 0.2, 0.3, 0.5, 1],
            labels=[
                "No Discount",
                "0–10%",
                "10–20%",
                "20–30%",
                "30–50%",
                "Above 50%"
            ]
        )
    )
    .groupby("Discount_Group", observed=False)
    .agg(
        Orders=("Profit", "count"),
        Total_Sales=("Sales", "sum"),
        Total_Profit=("Profit", "sum"),
        Average_Profit=("Profit", "mean"),
        Loss_Orders=("Profit", lambda x: (x < 0).sum())
    )
    .reset_index()
)

discount_analysis["Loss_Order_Percentage"] = (
    discount_analysis["Loss_Orders"]
    / discount_analysis["Orders"].replace(0, np.nan)
    * 100
)

display(discount_analysis)
```

```python

plt.figure(figsize=(10, 6))

sns.boxplot(
    data=df,
    x="Discount",
    y="Profit"
)

plt.title("Profit Distribution Across Discount Levels")
plt.xlabel("Discount Rate")
plt.ylabel("Profit")
plt.tight_layout()
plt.show()
```

```python

plt.figure(figsize=(9, 5))

sns.barplot(
    data=category_analysis,
    x="Category",
    y="Total_Profit"
)

plt.title("Total Profit by Product Category")
plt.xlabel("Product Category")
plt.ylabel("Total Profit")
plt.tight_layout()
plt.show()
```

```python

from pathlib import Path

output_dir = Path("visualizations")
output_dir.mkdir(exist_ok=True)

fig, ax = plt.subplots(figsize=(9, 5))
sns.barplot(
    data=category_analysis,
    x="Category",
    y="Total_Profit",
    ax=ax
)
ax.set_title("Total Profit by Product Category")
ax.set_xlabel("Product Category")
ax.set_ylabel("Total Profit")
fig.tight_layout()
fig.savefig(output_dir / "category_profit.png", dpi=300)
plt.close(fig)

fig, ax = plt.subplots(figsize=(9, 5))
sns.barplot(
    data=region_analysis,
    x="Region",
    y="Total_Profit",
    ax=ax
)
ax.set_title("Total Profit by Region")
ax.set_xlabel("Region")
ax.set_ylabel("Total Profit")
fig.tight_layout()
fig.savefig(output_dir / "region_profit.png", dpi=300)
plt.close(fig)

discount_analysis.to_csv("discount_profit_analysis.csv", index=False)
category_analysis.to_csv("category_analysis.csv", index=False)
region_analysis.to_csv("region_analysis.csv", index=False)

print("Charts and analysis files saved successfully.")
```

```python

worst_discount_group = discount_analysis.loc[
    discount_analysis["Average_Profit"].idxmin()
]

best_category = category_analysis.loc[
    category_analysis["Total_Profit"].idxmax()
]

worst_category = category_analysis.loc[
    category_analysis["Total_Profit"].idxmin()
]

print("Business Finding: Discount Analysis")
print(
    "Discount group with lowest average profit:",
    worst_discount_group["Discount_Group"]
)
print(
    "Average profit in this group:",
    round(worst_discount_group["Average_Profit"], 2)
)

print("\nCategory with highest total profit:")
print(best_category[["Category", "Total_Profit"]])

print("\nCategory with lowest total profit:")
print(worst_category[["Category", "Total_Profit"]])
```
